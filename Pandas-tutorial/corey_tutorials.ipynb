```python
country_grp = df.groupby(['country'])
df['country'].value_counts(normalize=True)
```

```python
country_grp['salary'].agg(['median','mean'])
```

```python
country_uses_python = country_grp['languageworkedwith'].apply(lambda x: (x.str.contains('Python').sum()))
country_uses_python
```

```python
country_respondents = df['country'].value_counts()
country_respondents
```

```python
country_uses_python/country_respondents
```

```python
python_df = pd.concat([country_uses_python,country_respondents],axis='columns',sort=False)
python_df
```

```python
python_df.rename(columns={'languageworkedwith':'KnowsPython','count':'Respondants'},inplace=True)
```

```python
python_df['pctknowspython'] = round((((python_df['KnowsPython'])/python_df['Respondants'])*100),2)
```

```python
python_df.sort_values('pctknowspython',ascending=False)
```
